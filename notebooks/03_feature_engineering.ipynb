## ETAPA 4 - Creación de nuevas features 
### Objetivo 
En este notebook se crearán nuevas variables derivadas a partir del dataset limpio generado en la fase anterior. Estas nuevas features permitirán representar mejor la información y preparar los datos para etapas posteriores de análisis o modelado.

### Regla de trabajo 
El archivo limpio ubicado en "C:\Proyecto-Risk-Score-OECE\data\processed\CONOSCE_ADJUDICACIONES_2018_2026_CONSOLIDADO_LIMPIO.csv" no será modificado directamente. Las nuevas variables se crearán sobre una copia de trabajo y el resultado se guardará como un nuevo archivo en data/processed/.
Parte

```python
# ==========================================
# FASE 4
# CREACIÓN DE NUEVAS FEATURES - RISK SCORE OSCE
# ==========================================

import pandas as pd
import numpy as np

# 1. Definir rutas de trabajo
# Ajusta estas rutas según tu carpeta real en Windows
ruta_limpio = r"C:\Proyecto-Risk-Score-OECE\data\processed\CONOSCE_ADJUDICACIONES_2018_2026_CONSOLIDADO_LIMPIO.csv"
ruta_featured = r"C:\Proyecto-Risk-Score-OECE\data\processed\CONOSCE_ADJUDICACIONES_2018_2026_CONSOLIDADO_FEATURED.csv"

# 2. Cargar el dataset limpio
# IMPORTANTE: el archivo usa punto y coma como separador
df_limpio = pd.read_csv(ruta_limpio, sep=";")

# 3. Crear copia de trabajo
df_features = df_limpio.copy()

# 4. Mostrar información inicial
print("Dimensión del dataset limpio:", df_features.shape)
print("\nColumnas originales:")
print(df_features.columns.tolist())
print("\nPrimeras filas:")
display(df_features.head())
```

```python
# ======================================================
# 5. CONVERTIR FECHAS A FORMATO DATETIME
# ======================================================

# FECHA_INICIO viene como número tipo 20200616
df_features["FECHA_INICIO_DT"] = pd.to_datetime(
    df_features["FECHA_INICIO"].astype(str),
    format="%Y%m%d",
    errors="coerce"
)

# FECHA_FIN puede venir con decimales o nulos, por ejemplo 20230716.0
df_features["FECHA_FIN_DT"] = pd.to_datetime(
    df_features["FECHA_FIN"].dropna().astype(float).astype(int).astype(str),
    format="%Y%m%d",
    errors="coerce"
)

# Reasignamos respetando los índices originales
fecha_fin_convertida = pd.to_datetime(
    df_features.loc[df_features["FECHA_FIN"].notna(), "FECHA_FIN"].astype(float).astype(int).astype(str),
    format="%Y%m%d",
    errors="coerce"
)

df_features["FECHA_FIN_DT"] = pd.NaT
df_features.loc[df_features["FECHA_FIN"].notna(), "FECHA_FIN_DT"] = fecha_fin_convertida

print("Fechas convertidas correctamente:")
display(df_features[["FECHA_INICIO", "FECHA_INICIO_DT", "FECHA_FIN", "FECHA_FIN_DT"]].head(10))
```

```python
df_features.to_csv(ruta_featured, sep=";", index=False, encoding="utf-8-sig")
```

```python
# ======================================================
# 5. CONVERTIR FECHAS A FORMATO DATETIME
# ======================================================

df_features["FECHA_INICIO_DT"] = pd.to_datetime(
    df_features["FECHA_INICIO"].astype(str),
    format="%Y%m%d",
    errors="coerce"
)

df_features["FECHA_FIN_DT"] = pd.to_datetime(
    df_features["FECHA_FIN"].astype(str).str.replace(".0", "", regex=False),
    format="%Y%m%d",
    errors="coerce"
)

print("Fechas convertidas:")
display(df_features[["FECHA_INICIO", "FECHA_INICIO_DT", "FECHA_FIN", "FECHA_FIN_DT"]].head())
```

```python
# ======================================================
# 6. CREAR FEATURES TEMPORALES
# ======================================================

df_features["ANIO_INICIO_SANCION"] = df_features["FECHA_INICIO_DT"].dt.year
df_features["MES_INICIO_SANCION"] = df_features["FECHA_INICIO_DT"].dt.month
df_features["TRIMESTRE_INICIO_SANCION"] = df_features["FECHA_INICIO_DT"].dt.quarter

df_features["PERIODO_INICIO_SANCION"] = (
    df_features["ANIO_INICIO_SANCION"].astype("Int64").astype(str)
    + "-"
    + df_features["MES_INICIO_SANCION"].astype("Int64").astype(str).str.zfill(2)
)

print("Features temporales creadas:")
display(df_features[[
    "FECHA_INICIO_DT",
    "ANIO_INICIO_SANCION",
    "MES_INICIO_SANCION",
    "TRIMESTRE_INICIO_SANCION",
    "PERIODO_INICIO_SANCION"
]].head())
```

```python
# ======================================================
# 7. CREAR FEATURES DE DURACIÓN
# ======================================================

df_features["DURACION_MESES"] = df_features["duracion_dias"] / 30
df_features["DURACION_ANIOS"] = df_features["duracion_dias"] / 365

df_features["SANCION_MAYOR_1_ANIO"] = np.where(
    df_features["duracion_dias"] > 365,
    1,
    0
)

df_features["SANCION_MAYOR_3_ANIOS"] = np.where(
    df_features["duracion_dias"] > 1095,
    1,
    0
)

df_features["LOG_DURACION_DIAS"] = np.log1p(df_features["duracion_dias"])

print("Features de duración creadas:")
display(df_features[[
    "duracion_dias",
    "DURACION_MESES",
    "DURACION_ANIOS",
    "SANCION_MAYOR_1_ANIO",
    "SANCION_MAYOR_3_ANIOS",
    "LOG_DURACION_DIAS"
]].head())
```

```python
print("Valores mínimos de duracion_dias:")
print(df_features["duracion_dias"].min())

print("\nCantidad de registros con duración negativa:")
print((df_features["duracion_dias"] < 0).sum())

print("\nCantidad de registros con duración igual a -1:")
print((df_features["duracion_dias"] == -1).sum())

print("\nRegistros problemáticos:")
display(df_features[df_features["duracion_dias"] <= 0][[
    "RUC",
    "FECHA_INICIO",
    "FECHA_FIN",
    "duracion_dias"
]].head(20))
```

```python
# ======================================================
# 7. CREAR FEATURES DE DURACIÓN
# ======================================================

df_features["DURACION_MESES"] = df_features["duracion_dias"] / 30
df_features["DURACION_ANIOS"] = df_features["duracion_dias"] / 365

df_features["SANCION_MAYOR_1_ANIO"] = np.where(
    df_features["duracion_dias"] > 365,
    1,
    0
)

df_features["SANCION_MAYOR_3_ANIOS"] = np.where(
    df_features["duracion_dias"] > 1095,
    1,
    0
)

# Corrección para evitar advertencias en el logaritmo
df_features["DURACION_DIAS_CORREGIDA"] = df_features["duracion_dias"].where(
    df_features["duracion_dias"] >= 0,
    np.nan
)

df_features["LOG_DURACION_DIAS"] = np.log1p(df_features["DURACION_DIAS_CORREGIDA"])

print("Features de duración creadas:")
display(df_features[[
    "duracion_dias",
    "DURACION_DIAS_CORREGIDA",
    "DURACION_MESES",
    "DURACION_ANIOS",
    "SANCION_MAYOR_1_ANIO",
    "SANCION_MAYOR_3_ANIOS",
    "LOG_DURACION_DIAS"
]].head())
```

```python
# ======================================================
# 8. CREAR FEATURES DEL RUC
# ======================================================

df_features["RUC_STR"] = df_features["RUC"].astype(str).str.zfill(11)
df_features["RUC_PREFIJO"] = df_features["RUC_STR"].str[:2]

df_features["ES_PERSONA_NATURAL"] = np.where(
    df_features["RUC_PREFIJO"] == "10",
    1,
    0
)

df_features["ES_PERSONA_JURIDICA"] = np.where(
    df_features["RUC_PREFIJO"] == "20",
    1,
    0
)

df_features["ES_OTRO_TIPO_RUC"] = np.where(
    ~df_features["RUC_PREFIJO"].isin(["10", "20"]),
    1,
    0
)

print("Features de RUC creadas:")
display(df_features[[
    "RUC",
    "RUC_STR",
    "RUC_PREFIJO",
    "ES_PERSONA_NATURAL",
    "ES_PERSONA_JURIDICA",
    "ES_OTRO_TIPO_RUC"
]].head())
```

```python
# ======================================================
# 9. CREAR FEATURES DEL MOTIVO DE INFRACCIÓN
# ======================================================

df_features["MOTIVO_TEXTO"] = df_features["DE_MOTIVO_INFRACCION"].fillna("").str.lower()

df_features["MOTIVO_DOC_FALSA"] = np.where(
    df_features["MOTIVO_TEXTO"].str.contains("fals|adulter", regex=True),
    1,
    0
)

df_features["MOTIVO_INFO_INEXACTA"] = np.where(
    df_features["MOTIVO_TEXTO"].str.contains("inexacta", regex=False),
    1,
    0
)

df_features["MOTIVO_INCUMPLIMIENTO"] = np.where(
    df_features["MOTIVO_TEXTO"].str.contains("incumpl", regex=True),
    1,
    0
)

df_features["MOTIVO_CONTRATAR_IMPEDIDO"] = np.where(
    df_features["MOTIVO_TEXTO"].str.contains("impedid|impedimento", regex=True),
    1,
    0
)

df_features["NUM_MOTIVOS_INFRACCION"] = (
    df_features["ID_MOTIVO_INFRACCION"]
    .fillna("")
    .astype(str)
    .str.split(",")
    .apply(lambda lista: len([x for x in lista if x.strip() != ""]))
)

df_features["TIENE_MULTIPLES_MOTIVOS"] = np.where(
    df_features["NUM_MOTIVOS_INFRACCION"] > 1,
    1,
    0
)

print("Features de motivo de infracción creadas:")
display(df_features[[
    "DE_MOTIVO_INFRACCION",
    "MOTIVO_DOC_FALSA",
    "MOTIVO_INFO_INEXACTA",
    "MOTIVO_INCUMPLIMIENTO",
    "MOTIVO_CONTRATAR_IMPEDIDO",
    "NUM_MOTIVOS_INFRACCION",
    "TIENE_MULTIPLES_MOTIVOS"
]].head())
```

```python
# ======================================================
# 10. GUARDAR ARCHIVO FEATURED
# ======================================================

df_features.to_csv(
    ruta_featured,
    sep=";",
    index=False,
    encoding="utf-8-sig"
)

print("Archivo con nuevas features guardado correctamente en:")
print(ruta_featured)
```

```python
# ======================================================
# 10. CREAR FEATURES HISTÓRICAS POR PROVEEDOR
# ======================================================

# Ordenamos por RUC y fecha de inicio de sanción
df_features = df_features.sort_values(["RUC", "FECHA_INICIO_DT"])

# Número histórico de sanciones del proveedor en la base
df_features["TOTAL_SANCIONES_PROVEEDOR"] = df_features.groupby("RUC")["RUC"].transform("count")

# Orden de la sanción dentro del historial del proveedor
df_features["ORDEN_SANCION_PROVEEDOR"] = df_features.groupby("RUC").cumcount() + 1

# Indicador de si el proveedor aparece más de una vez en la base
df_features["PROVEEDOR_CON_MULTIPLES_SANCIONES"] = np.where(
    df_features["TOTAL_SANCIONES_PROVEEDOR"] > 1,
    1,
    0
)

# Fecha de primera sanción del proveedor
df_features["FECHA_PRIMERA_SANCION_PROVEEDOR"] = df_features.groupby("RUC")["FECHA_INICIO_DT"].transform("min")

# Días desde la primera sanción observada
df_features["DIAS_DESDE_PRIMERA_SANCION"] = (
    df_features["FECHA_INICIO_DT"] - df_features["FECHA_PRIMERA_SANCION_PROVEEDOR"]
).dt.days

print("Features históricas por proveedor creadas:")
display(df_features[[
    "RUC",
    "FECHA_INICIO_DT",
    "TOTAL_SANCIONES_PROVEEDOR",
    "ORDEN_SANCION_PROVEEDOR",
    "PROVEEDOR_CON_MULTIPLES_SANCIONES",
    "FECHA_PRIMERA_SANCION_PROVEEDOR",
    "DIAS_DESDE_PRIMERA_SANCION"
]].head(20))
```

```python
# ======================================================
# 11. IDENTIFICAR TARGET Y VARIABLES QUE NO DEBEN USARSE COMO FEATURES
# ======================================================

target = "target_sancionado_24m"

# Estas columnas no deben usarse como predictoras en el modelo,
# porque contienen información futura o identificadores directos.
columnas_no_modelo = [
    "target_sancionado_24m",
    "fecha_segunda",                 # fuga de información
    "FECHA_INICIO_DT",
    "FECHA_FIN_DT",
    "FECHA_PRIMERA_SANCION_PROVEEDOR",
    "MOTIVO_TEXTO",
    "NOMBRE_RAZONODENOMINACIONSOCIAL",
    "NUMERO_RESOLUCION",
    "RUC_STR"
]

print("Variable objetivo:", target)
print("\nColumnas que NO deben usarse como features predictoras:")
print(columnas_no_modelo)
```

```python
# ======================================================
# 12. MOSTRAR EVIDENCIA DE LAS NUEVAS FEATURES
# ======================================================

print("\nDimensión después de crear features:", df_features.shape)

print("\nColumnas después de crear nuevas features:")
print(df_features.columns.tolist())

print("\nDistribución del target:")
print(df_features[target].value_counts())

print("\nDistribución porcentual del target:")
print(df_features[target].value_counts(normalize=True) * 100)

print("\nVista previa de features principales:")
display(df_features[[
    "RUC",
    "FECHA_INICIO_DT",
    "ANIO_INICIO_SANCION",
    "MES_INICIO_SANCION",
    "duracion_dias",
    "DURACION_MESES",
    "SANCION_MAYOR_1_ANIO",
    "ES_PERSONA_NATURAL",
    "ES_PERSONA_JURIDICA",
    "MOTIVO_DOC_FALSA",
    "MOTIVO_INFO_INEXACTA",
    "NUM_MOTIVOS_INFRACCION",
    "TOTAL_SANCIONES_PROVEEDOR",
    "PROVEEDOR_CON_MULTIPLES_SANCIONES",
    target
]].head(20))

# 13. Guardar el nuevo archivo con features
df_features.to_csv(ruta_featured, sep=";", index=False, encoding="utf-8-sig")

print("\nArchivo con nuevas features guardado en:")
print(ruta_featured)
```
