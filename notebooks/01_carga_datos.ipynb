```python
# ==========================================
# PUNTO 4: CARGAR EL CSV DESDE data/raw/
# ==========================================

# ------------------------------------------------------------------
# PASO 1: Importar pandas
# ------------------------------------------------------------------

# Importamos pandas, la librería principal para trabajar con datos tabulares
# La abreviatura 'pd' es un estándar en la comunidad de Data Science
import pandas as pd

# ------------------------------------------------------------------
# PASO 2: Definir la ruta del archivo
# ------------------------------------------------------------------

# Guardamos la ubicación del archivo en una variable llamada 'ruta_csv'
# Usamos una cadena cruda (r"...") para que Python no interprete las \ como caracteres especiales
# Esta es una ruta absoluta: comienza desde la raíz C:\
ruta_csv = r"C:\Proyecto-Risk-Score-OECE\data\raw\CONOSCE_ADJUDICACIONES_2018_2026_CONSOLIDADO.csv"

# ------------------------------------------------------------------
# PASO 3: Cargar el archivo CSV
# ------------------------------------------------------------------

# Leemos el archivo CSV y lo convertimos en un DataFrame (tabla)
# El DataFrame se llama 'df' por convención (abreviatura de DataFrame)
# 
# PARÁMETROS IMPORTANTES:
# - sep=';' : el archivo usa punto y coma como separador (no coma)
# - encoding='utf-8' : permite leer correctamente letras con acentos y eñes
df = pd.read_csv(ruta_csv, sep=';', encoding='utf-8')

# ------------------------------------------------------------------
# PASO 4: Verificar la carga mostrando las primeras filas
# ------------------------------------------------------------------

# El método head() muestra las primeras 5 filas del DataFrame
# Usamos print() para que el resultado sea visible en el notebook
# Esto nos permite confirmar que el archivo se cargó correctamente
print("\n" + "="*100)
print("PRIMERAS 100 FILAS DEL DATASET:")
print("="*50)
print(df.head())
```
